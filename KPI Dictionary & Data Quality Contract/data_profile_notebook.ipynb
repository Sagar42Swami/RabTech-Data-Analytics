# Retail Data Quality Profile
Profiles the supplied raw retail orders against the supplied data dictionary.

```python
from pathlib import Path
import pandas as pd
base=Path.cwd()
data_path=base/'retail-orders-raw (1).csv'
df=pd.read_csv(data_path)
df.head()
```

```python
profile=pd.DataFrame({'column':df.columns,'missing_count':df.isna().sum().values,'unique_count':df.nunique(dropna=True).values,'dtype':df.dtypes.astype(str).values})
profile
```

```python
dates=pd.to_datetime(df['order_date'],errors='coerce')
quantity=pd.to_numeric(df['quantity'],errors='coerce')
discount=pd.to_numeric(df['discount_pct'],errors='coerce')
checks={
'duplicate_order_rows':int(df['order_id'].duplicated(keep=False).sum()),
'invalid_dates':int(dates.isna().sum()),
'invalid_quantity':int((quantity.isna()|(quantity<=0)).sum()),
'invalid_discount':int(((discount<0)|(discount>100)).sum()),
'missing_required_cells':int(df.isna().sum().sum())}
pd.Series(checks,name='count')
```

## Contract interpretation
Any Critical rule failure blocks publication of affected sales KPIs. The raw CSV remains unchanged.
